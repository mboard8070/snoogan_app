```python
from polygon import RESTClient
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.getenv("POLYGON_API_KEY")

print("API Key exists:", bool(api_key))
if api_key:
    print("Key preview:", api_key[:10] + "..." if len(api_key) > 10 else "short")

client = RESTClient(api_key=api_key)

# Test 1: Can we reach Polygon at all?
try:
    test = client.get_last_trade("AAPL")
    print("Polygon connection: SUCCESS")
    print("AAPL last price:", test.price)
except Exception as e:
    print("Polygon connection: FAILED")
    print("Error:", str(e))

# Test 2: Can we get SPX price?
try:
    from datetime import date, timedelta
    aggs = client.get_aggs("SPX", 1, "day", from_=(date.today() - timedelta(days=5)), to=date.today())
    print("SPX aggs success:", len(aggs) > 0)
except Exception as e:
    print("SPX aggs failed:", str(e))
```
