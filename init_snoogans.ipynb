```python
import os
import sys

# Print kernel's starting point (probably some /project bullshit)
print("Kernel starting cwd:", os.getcwd())
print("Starting dir contents:", os.listdir('.')[:10])

# Auto-find the project root by lookin' for code/ and data/
current = os.getcwd()
project_root = None
for _ in range(10):  # Climb up max 10 levels
    contents = os.listdir(current) if os.path.exists(current) else []
    if 'code' in contents and 'data' in contents and os.path.exists(os.path.join(current, 'code', 'rag')):
        project_root = current
        break
    parent = os.path.dirname(current)
    if parent == current:  # Hit filesystem root
        break
    current = parent

if project_root is None:
    raise ValueError("Snoogans can't find project root! Need 'code' and 'data' folders. Check Workbench mounts.")

# Lock in the root
os.chdir(project_root)
if project_root not in sys.path:
    sys.path.insert(0, project_root)

print("\nLocked project root:", os.getcwd())
print("sys.path[0]:", sys.path[0])
print("code/ exists?", os.path.exists("code"))
print("code/rag/snoogans_brain.py exists?", os.path.exists("code/rag/snoogans_brain.py"))
print("Files in code/rag:", os.listdir("code/rag"))
print("data/knowledge exists?", os.path.exists("data/knowledge"))
```
